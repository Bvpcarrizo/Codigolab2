```python
tm=np.array([1.78,1.81,1.72,1.40,1.39,1.41,0.96,1.07,0.98,1.00])
theta=np.array([0.07,0.075,0.083,0.087,0.1,0.104,0.108,0.122,0.157,0.174])
x = 1 / np.sin(theta)
y = tm**2
coef = np.polyfit(x, y, 1)
m = coef[0]
b = coef[1]
y_fit = np.polyval(coef, x)
residuos = y - y_fit
RSS = np.sum(residuos**2)
SS_tot = np.sum((y - np.mean(y))**2)
R2 = 1 - RSS / SS_tot
print("Pendiente =", m)
print("Intercepto =", b)
print("RSS =", RSS)
print("R^2 =", R2)
xx = np.linspace(np.min(x),np.max(x),200)
yy = np.polyval(coef,xx)
plt.figure()
plt.scatter(x,y,label="Datos experimentales")
plt.plot(xx,yy,label="Ajuste lineal")
plt.errorbar(x,y, yerr = D_t , xerr = D_a, fmt='none')
plt.xlabel(r"$1/\sin(\theta)$")
plt.ylabel(r"$t^2$ [s$^2$]")
plt.grid(True)
plt.legend()
plt.savefig( "regresióngrav.pdf", bbox_inches="tight")
plt.show()
```
